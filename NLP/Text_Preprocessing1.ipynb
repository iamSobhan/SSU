**Text Preprocessing**<br><br>

<p>Text preprocessing is the process of cleaning and transforming raw text into a structured format so that it can be easily understood and processed by computers in Natural Language Processing.<br>
Text Preprocessing includes, <br>
1. Text Cleaning <br>
2. Text Lowercasing <br>
3. Tokenization <br>
4. Stopword Removal <br>
5. Removing Punctuation and Special Characters <br>
6. Stemming <br>
7. Lemmatization <br>
and others.


**1. Text Cleaning** <br>

<p>Text cleaning is the first and most foundational step in any NLP pipeline. It is the process of removing, correcting, or normalizing parts of raw text that add noise, inconsistency, or irrelevance — leaving behind only the meaningful content that a model can learn from effectively.</p>

**1.2. Remove URLs** <br>
URL removal is a text cleaning step in Natural Language Processing where web links (URLs) are removed from text.

**Why Removing**<br>
Removes unnecessary links from text.<br>
Makes text cleaner and easier to process.<br>
Improves NLP model performance.<br>
Useful for social media and web data cleaning.<br>

```python
import re

#Input text
text = """Visit https://google.com for more information. NLP tutorials are available at www.nlpstudy.com"""

#Removing URLs
clean_text = re.sub(r'https?://\S+|www\.\S+', '', text)

print(clean_text)
```

**Advantages**<br>
Removes noisy text.<br>
Cleans web and social media data.<br>
Improves text preprocessing quality.<br>
Makes analysis more meaningful.

**1.3. Removing Punctuation and Special Characters** <br>
<p>Removing punctuation and special characters is a text cleaning step in Natural Language Processing where symbols such as:. , ! ? ; : @ # $ % & * ( ) [ ] { } / \ | + = < > are removed from text. These characters often do not contribute meaningful information to NLP tasks.</p>

**Why Removing**<br>
Cleans noisy text.<br>
Improves text quality.<br>
Reduces unnecessary tokens.<br>
Makes text easier to analyze.<br>
Improves machine learning performance.

```python
import string

text = "Hello!!! Welcome to Seacom Skills University @2026."

#Removing punctuation
clean_text = text.translate(
    str.maketrans('', '', string.punctuation)
)

print(clean_text)
```

```python
import re

text = "Hello!!! Welcome to the department of Computer Science and Engineering @2026."

clean_text = re.sub(r'[^a-zA-Z0-9\s]', '', text)

print(clean_text)
```

```python
import re

text = """
Hii, there!!! Welcome to #West Bengal @2026$$$. West Bengal is #the best #state #1 state in Tourism.
"""

print("Original Text:")
print(text)

clean_text = re.sub(r'[^a-zA-Z0-9\s]', '', text)

print("\nAfter Removing Punctuation & Special Characters:")
print(clean_text)
```

**Advantages**<br>
Removes noisy symbols.<br>
Improves tokenization.<br>
Reduces unnecessary features.<br>
Makes text cleaner for NLP.

**1.4. Normalize Whitespace** <br>
<p>Whitespace normalization is a text cleaning step in Natural Language Processing that removes unnecessary: Multiple spaces, Tabs (\t), New lines (\n), Extra blank spacesand converts them into a single space.</p>

**Why Normalize**<br>
<p>During data collection from Websites, PDFs, Social media, Documents, text often contains irregular spaces. This makes text difficult to process. Whitespace normalization solve this.</p>

```python
import re

text = """
Hello      World

Welcome      to NLP.
"""

# Normalize whitespace
clean_text = re.sub(r'\s+', ' ', text).strip()

print(clean_text)
```

```python
import re

text = "NLP\t\tis\n\nvery     interesting"

clean_text = re.sub(r'\s+', ' ', text).strip()

print(clean_text)
```

```python
text = """
Hello      World

Welcome      to NLP
"""

clean_text = " ".join(text.split())

print(clean_text)
```

**Advantages**<br>
Removes unwanted spaces.<br>
Improves tokenization.<br>
Makes text uniform.<br>
Reduces noise in datasets,<br>
Improves NLP preprocessing quality.

**1.5. Remove Numbers**<br>
Number removal is a text cleaning step in Natural Language Processing where numerical values are removed from text.

**Why Removing**<br>
Removes irrelevant information.<br>
Simplifies text analysis.<br>
Reduces noise in datasets.<br>
Improves text preprocessing.<br>
Useful in sentiment analysis and text classification.

```python
import re

text = "I bought 5 books in 2026 and spent 1500 rupees."

# Remove numbers
clean_text = re.sub(r'\d+', '', text)

print(clean_text)
```

```python
import re

text = """
My phone number is 9876543210.
I scored 90 marks.
"""

clean_text = re.sub(r'\d+', '', text)

print(clean_text)
```

```python
import re

text = "The price is 45.99 rupees."

clean_text = re.sub(r'\d+(\.\d+)?', '', text)

print(clean_text)
```

**Advantages**<br>
Removes unwanted numerical noise.<br>
Simplifies text processing.<br>
Reduces feature size.<br>
Useful for text classification.

**1.6. Handle Emojis** <br>
Emoji handling is a text cleaning step in Natural Language Processing where emojis are either:<br>
Removed from the text, or <br>
Converted into meaningful words. <br>
Emojis often contain important emotional information and are widely used in social media, reviews, chats, and tweets.

**Why Handling**<br>
Emojis can affect sentiment analysis.<br>
Improves text preprocessing.<br>
Helps understand emotions in text.<br>
Makes text machine-readable.

```python
import re

text = "I love Coding 😊🚀🔥"

#Removing emojis
clean_text = re.sub(r'[^\x00-\x7F]+', '', text)   #r'[^\x00-\x7F]+' --> this for removing all non-ASCII characters. Emojis are non-ASCII Unicode characters.

print(clean_text)
```

```python
import emoji

text1 = "I am very happy 😊"
text2 = "The movie was amazing 😍🔥❤️"
text3 = "I love Coding 😊🚀🔥"

print(emoji.demojize(text1))

print(emoji.demojize(text2))

print(emoji.demojize(text3))
```

**Advantages of Emoji Conversion**<br>
Preserves sentiment information.<br>
Useful for social media analytics.<br>
Better than deleting emojis.<br>
Helps emotion detection models.<br><br>

**Advantages of Emoji Removal**

Simpler preprocessing.<br>
Reduces noise.<br>
Useful when emojis are irrelevant.
